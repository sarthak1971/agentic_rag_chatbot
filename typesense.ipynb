```python
### Langchain + Typsense + Groq LLM + RAG Application

from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import Typesense
from langchain_text_splitters import CharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_groq import ChatGroq
```

```python
import os
os.environ["GROQ_API_KEY"]
```

```python
loader = TextLoader("test.txt")
documents = loader.load()
text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap=100)
docs = text_splitter.split_documents(documents)

embeddings = HuggingFaceEmbeddings()
```

```python
docsearch=Typesense.from_documents(
    docs,
    embeddings,
    typesense_client_params={
        "host": "4apqxcglysn79or8p-1.a1.typesense.net",  # Use xxx.a1.typesense.net for Typesense Cloud
        "port": "443",  # Use 443 for Typesense Cloud
        "protocol": "https",  # Use https for Typesense Cloud
        "typesense_api_key":"iCQySFnMq6tAhWLduLvJ1btgDHvzWHRj",
        "typesense_collection_name": "lang-chain"
    },
)
```

```python
query = "What is Python"
found_docs = docsearch.similarity_search(query)
print(found_docs[0].page_content)
```

```python
### Retriever
retriever = docsearch.as_retriever()
retriever
```
